### Data Structure

```python
#List
List=[21,34,56,"list"]
print(type(List))
print(List[3])
print(List[-1])
List[2]="Array"
print(List[2])
List.append("orange")
List.insert(3,"banana")
List.remove("list")
List.pop(1)
List.pop()  #remove the list item of list
List.clear() #remove all the list items and return an empty list
print(len(List))
print(List)
```
