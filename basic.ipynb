## DataTypes

### Numbers

```python
#int
a=2

#float
b=2.5

#complex
c=2+3j


print(type(a))
print(type(b))
print(type(c))
```

### Boolean

```python
# true and false
a=False
b=True
print(type(b))
print(type(a))
print(10<6)
```

### strings

```python
a="Hammer"

#datatype
print(type(a))

#indexing
print(a[0])
print(a[1]) 

#slicing
print(a[1:])
print(a[ :3])

#upper case
print(a.upper())    

#lower case
print(a.lower()) 

#replace values
print(a.replace("Hammer","nails"))
```

### Type Conversion

```python
#Implicit

a=5
b=3+3j
c=a+b
print(type(c))
print(c)

#Explicit
i=56
d="23"
e=int(d)
f=bool(e)
g=float(f)
print(type(d))
print(type(e))
print(type(f))
print(type(g))

print(i+e)
```
