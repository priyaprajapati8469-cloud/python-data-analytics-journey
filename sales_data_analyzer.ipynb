### Reading the file

```python
file=open("sales_data.csv","r")
r=file.readlines()              #readline-read only first row     readlines-read entire data
print(r)
```

```python
newdata=[]
for line in r [1: ]:
    field = line.strip().split(",")     #strip-removes leading space "  priya"  and trailing spaces "priya  "
    field[5]=int(field[5])             #coverting 5 and 6 row into integer
    field[6]=int(field[6])
    newdata.append(field)

print(newdata)    
```

### 1.what is the total revenue from total sales

```python
total_revenue = 0
for row in newdata:
    quantity=row[5]
    price=row[6]
    total_revenue+=quantity*price
print("total Revenue:",total_revenue)
```

### 3.how many total items (quantity) were sold

```python
total_quantity=0
for row in newdata:
 total_quantity+=row[5]
print("total qantity sold:",total_quantity)    
```

### 4. total revenue  generated by each category?

```python
category_revenue={}
for row in newdata:
    category=row[4]
    revenue=row[5]*row[6]
    if category in category_revenue:
        category_revenue[category]+=revenue
    else:
         category_revenue[category]=revenue  
  
for cat,rev in category_revenue.items():
    print(f"{cat} : {rev}")        
```

### 5.what is the total revenue generated bye each product


```python
product_revenue={}
for row in newdata:
    product=row[3]
    revenue=row[5]*row[6]
    if product in product_revenue:
        product_revenue[product]+=revenue
    else:
         product_revenue[product]=revenue  
# print("revenue by product",product_revenue)  
for pro,rev in product_revenue.items():
    print(f"{pro} : {rev}")   
```

### 6.which product has the highest quantity sold

```python
product_quantity={}
for row in newdata:
    product=row[3]
    quantity=row[5]
if  product in product_quantity:
    product_quantity[product]+=quantity
else:
    product_quantity[product]=quantity 

highest_sold=None
max_quantity=0

for prod,qty in product_quantity.items():
    if qty>max_quantity:
        max_quantity=qty
        highest_sold=prod

print("highest product sold:",highest_sold ,"\nmax quantity:",max_quantity)        


```

### 7.which customer spends the most money?

```python
customer_revenue={}
for row in newdata:
    customer=row[2]
    revenue=row[5]*row[6]

    if customer in customer_revenue:
        customer_revenue[customer]+=revenue
    else:
        customer_revenue[customer]=revenue   
top_customer=None
max_revenue=0
for cust,rev in customer_revenue.items():
    if rev>max_revenue:
        max_revenue=rev
        top_customer=cust
print("top customer:",top_customer,"\nmax revenue:",max_revenue)                
```

### 8.what is the total revenue for each date?

```python
daily_revenue = {}
for row in newdata:
    date=row[1]
    revenue=row[5]*row[6]
if date in daily_revenue:
    daily_revenue[date] += revenue
else:
    daily_revenue[date] = revenue    
print("daily revenue") 
for date, rev in daily_revenue.items():
    print(f"{date}: {rev}")        
```

### 9. what is the total revenue between two given dates

```python
from datetime import datetime
start=input("enter start date date (YYYY-MM-DD)")
end=input("enter start end date (YYYY-MM-DD)")
start_date=datetime.strptime(start,"%Y-%M-%d")
end_date=datetime.strptime(end,"%Y-%M-%d")
total_daterev=0
for row in newdata:
    order_date=datetime.strptime(row[1],"%Y-%M-%d")
    if start_date<=order_date<=end_date:
     total_daterev +=row[5]*row[6]
print("revenue from", start,"and",end,"is",total_daterev)    
```

### 10. export all order under the "Electronics" category inton a new csv file 

```python
elecfile=open("electronics.csv","w")
elecfile.write("OrderID,Date,Customer,Product,Category,Quantity,Price\n")
for row in newdata:
   if row[4]=="Electronics":
      row_str=",".join([str(item) for item in row])
      elecfile.write(row_str + "\n")
elecfile.close()      

print("all electrical items saved to electrical.csv")
  
```
